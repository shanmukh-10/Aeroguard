```python
clean_df = df.copy()

print("Rows:", len(clean_df))
print("Columns:", len(clean_df.columns))
```

```python
clean_df["Timestamp"] = pd.to_datetime(
    clean_df["Timestamp"],
    errors="coerce"
)
```

```python
print(
    "Invalid timestamps:",
    clean_df["Timestamp"].isna().sum()
)
```

```python
print("Before removing invalid rows:", len(clean_df))

clean_df = clean_df.dropna(
    subset=["Timestamp"]
).copy()

print("After removing invalid rows:", len(clean_df))
```

```python
clean_df = clean_df.sort_values(
    "Timestamp"
).reset_index(drop=True)
```

```python
print(
    "Duplicate rows:",
    clean_df.duplicated().sum()
)
```

```python
missing_summary = pd.DataFrame({
    "missing_count": clean_df.isnull().sum(),
    "missing_percentage": (
        clean_df.isnull().sum() /
        len(clean_df) * 100
    )
})

missing_summary = missing_summary.sort_values(
    "missing_percentage",
    ascending=False
)

missing_summary
```

```python
print(
    "PM2.5 missing:",
    clean_df["PM2.5 (µg/m³)"].isna().sum()
)

print(
    "PM2.5 missing %:",
    clean_df["PM2.5 (µg/m³)"].isna().mean() * 100
)
```
